```python
import pandas as pd

df = pd.read_csv("../data/processed/telco_clean.csv")

df.head()

```

```python
analysis_cols = [
    "Age",
    "Tenure in Months",
    "Monthly Charge",
    "Total Charges",
    "Total Revenue",
    "Internet Type",
    "Number of Dependents",
    "Satisfaction Score",
    "Married",
    "Churn Label",
    "Churn Score"
]

df_analysis = df[analysis_cols]
df_analysis.head()
```

```python
revenue = df_analysis["Total Revenue"].dropna()

mean_revenue = revenue.mean()
sem_revenue = stats.sem(revenue)

confidence_level = 0.95
ci = stats.t.interval(
    confidence_level,
    df=len(revenue) -1,
    loc= mean_revenue,
    scale=sem_revenue
)

print("Mean Revenue:", round(mean_revenue, 2))
print("95% Confidence interval:",
      round(ci[0],2),
      "to",
      round(ci[1],2))
```

```python
group_0 = df_analysis[
    df_analysis["Number of Dependents"] == 0
]["Total Revenue"].dropna()

group_1 = df_analysis[
    df_analysis["Number of Dependents"] > 0
]["Total Revenue"].dropna()
```

```python
t_stat , p_value = stats.ttest_ind(
    group_0,
    group_1,
    equal_var=False
)

print("T-statistics", round(t_stat, 3))
print("P-value:", round(p_value, 4))
```

```python
if p_value < 0.05:
    print("Reject H0: There is a ststistically significant difference.")
else:
    print("Fail to reject H0: No statistically significant difference.")
```

```python
contigency_table = pd.crosstab(
    df_analysis["Married"],
    df_analysis["Churn Label"]
)

contigency_table
```

```python
from scipy import stats
chi2, p_value, dof, expected = stats.chi2_contingency(
    contigency_table
)

print("Chi-square statistics:", round(chi2, 3))
print("Degree of Freedom:", dof)
print("P-values", round(p_value, 4))
```

```python
print ("chi-square test result -")
if p_value < 0.05:
    print("Reject H0: There is significant association between marital status and churn.")
else:
    print("Fail to reject H0: There is no statistically significant association between marital status and churn.")
```

## Chi-Square Test: Marital Status vs Churn

### Hypotheses
- **H₀ (Null Hypothesis):** Marital status and churn are independent.
- **H₁ (Alternative Hypothesis):** Marital status and churn are associated.

### Significance Level
α = 0.05

### Interpretation
The Chi-square test was used to determine whether marital status is statistically associated with customer churn.

If the p-value is less than 0.05, we reject H₀ and conclude that there is a statistically significant association between marital status and churn.

If the p-value is greater than or equal to 0.05, we fail to reject H₀ and conclude that there is insufficient statistical evidence of an association between marital status and churn.

## Correlation vs Causation

Correlation measures the strength and direction of a relationship between two variables, but it does not establish that one variable causes the other.

For example, a positive correlation between customer tenure and total revenue would indicate that customers with longer tenure tend to have higher revenue. However, this does not prove that increasing tenure itself causes revenue to increase.

Other factors such as monthly charges, services subscribed to, customer segment, contract type, and other characteristics may influence revenue.

Therefore, correlation findings should be interpreted as associations rather than causal relationships.

- **H₀:** The mean Total Revenue is equal across all groups.
- **H₁:** At least one group has a different mean Total Revenue.

### Results
- **F-statistic:** 145.251
- **P-value:** < 0.001
- **Significance level:** 0.05

### Conclusion
Since the p-value is less than 0.05, we reject the null hypothesis. There is statistically significant evidence that average Total Revenue differs across the groups.

### Business Implication
The groups appear to have different revenue performance. This suggests that the group/category being analyzed may be associated with customer revenue and should be investigated further to identify which groups generate higher or lower revenue.

However, ANOVA only indicates that at least one group differs. A post-hoc test would be required to identify the specific groups responsible for the difference.

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
X_scaled[:5]
```

```python
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt
inertia =[]

for k in range(2,12):
    kmeans = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )
    kmeans.fit(X_scaled)
    inertia.append(kmeans.inertia_)
```

```python
plt.figure(figsize=(9,5))

plt.plot(range(2, 12),
        inertia,
        marker='o')

plt.title('Elbow Method')
plt.xlabel('Number of clusters (k)')
plt.ylabel('Inertia')

plt.show()
```

```python
from sklearn.metrics import silhouette_score

silhouette_scores = []

for k in range(2, 11):

    kmeans = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    labels = kmeans.fit_predict(X_scaled)
    score = silhouette_score(X_scaled, labels)
    silhouette_scores.append(score)
    print(f"k = {k}, Silhouette Score = {score:.4f}")
```

```python
plt.Figure(figsize=(10,5))

plt.plot(
    range(2,11),
    silhouette_scores,
    marker='o'
)

plt.title('Silhouette Score')
plt.xlabel('Number of Clusters (K)')
plt.ylabel('Silhouette Score')

plt.show()
```

```python
best_k = range(2,11)[
    silhouette_scores.index(max(silhouette_scores))
]
print("Best K:" , best_k)
```

```python
kmeans = KMeans(
    n_clusters=best_k,
    random_state=42,
    n_init=10
)
df['Cluster'] = kmeans.fit_predict(X_scaled)
```

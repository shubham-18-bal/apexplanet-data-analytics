```python
import matplotlib.pyplot as plt

age_churn =df.groupby ('Age')['Churn Score'].mean().reset_index()
plt.figure(figsize=(10,5))
plt.plot(
    age_churn['Age'],
    age_churn['Churn Score'],
    color='#2b5c8f',
    linewidth=2,
    marker='o',
    markersize=4
)
plt.title('Average churn score across customer age')
plt.xlabel('Age')
plt.ylabel('Average Churn Score')

plt.tight_layout()
plt.show()
```

```python
import numpy as np
grouped_data = df.groupby(['Senior Citizen','Gender'])['Total Revenue'].mean().unstack()

categories = grouped_data.index
x = np.arange(len(categories))
width =0.35


fig, ax = plt.subplots(figsize=(8, 5))

rects1 = ax.bar(x - width/2, grouped_data['Female'], label ='female')
rects2 = ax.bar(x + width/2, grouped_data['Male'], label = 'male')

ax.set_title('Average total revenue: senior citizen vs non-senior by gender')
ax.set_xlabel('Senior Citizen')
ax.set_ylabel('Average Total Revenue ($)')
ax.set_xticks(x)
ax.set_xticklabels(categories)
ax.legend(title='Gender', frameon=True)
ax.grid(axis='y' , linestyle='--', alpha=0.5)

ax.bar_label(rects1, fmt='%.1f', padding=1, fontsize=9)
ax.bar_label(rects2, fmt='%.1f', padding=1, fontsize=9)

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(9,5))

n, bins, patches = plt.hist(
    df['Total Revenue'].dropna(),
    bins=25,
)

mean_rev =df['Total Revenue'].mean()
plt.axvline(mean_rev, color='#d9534f')

plt.title('Distribution of customer total revenue', fontsize=13)
plt.xlabel('Total Revenue', fontsize=10)
plt.ylabel('Frequency (number of customers)', fontsize=11)
plt.legend(frameon=True)
plt.grid(axis='y')

plt.show()
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(9,6))

scatter = plt.scatter(
    df['Age'],
    df['Total Revenue'],
    c=df['Churn Score'],
    cmap='viridis',
    alpha=0.6,
    s=40
)

cbar = plt.colorbar(scatter)
cbar.set_label('Churn Score (Higher = more at risk)', fontsize=10)
plt.title('customer age VS total revenue (colored by churn score)', fontsize=13)
plt.xlabel('Age')
plt.ylabel('Tot. Revenue ($)')
plt.grid(True, linestyle='--')

plt.tight_layout()
```

```python
fig, axes = plt.subplots(nrows=2, ncols=2, figsize=(14,10))

axes[0,0].hist(df['Age'].dropna(), bins=20)
axes[0,0].set_title('Age Distribution')
axes[0,0].set_xlabel('Age')
axes[0,0].set_ylabel('Count')
axes[0,0].grid(axis='y')

sat_counts = df['Satisfaction Score'].value_counts().sort_index()
axes[0,1].bar(sat_counts.index, sat_counts.values, color='#4cae4c')
axes[0,1].set_title('Satisfaction Score breakdown')
axes[0,1].set_xlabel('Satisfaction score(1-5)')
axes[0,1].set_ylabel('Count')
axes[0,1].grid(axis='y')

axes[1,0].scatter(df['Age'], df['Total Revenue'])
axes[1,0].set_title('Age vs Total Revenue')
axes[1,0].set_xlabel('Age')
axes[1,0].set_ylabel('Total Rvn')
axes[1,0].grid(True)

cltv_gender = df.groupby('Gender')['CLTV'].mean()
axes[1,1].bar(cltv_gender.index, cltv_gender.values)
axes[1,1].set_title('Average CLTV by gender')
axes[1,1].set_xlabel('Gender')
axes[1,1].set_ylabel('Avg CLTV')
axes[1,1].grid(axis='y')


plt.suptitle('Customer overview analytics Dashboard')
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

numeric_cols = [
    'Age' , 'Number of Dependents' , 'Total Extra Data Charges' , 'Total Long Distance Charges' ,
    'Total Revenue' , 'Satisfaction Score' , 'Churn Score' , 'CLTV'
]

corr_matrix = df[numeric_cols].corr()

plt.figure(figsize=(10,8))
sns.heatmap(
    corr_matrix,
    annot=True,
    fmt="2f",
    cmap="coolwarm",
    cbar_kws={'label': 'Correlation Coefficient'}
)

plt.title('Feature Correlation Matrix')
plt.xticks(rotation=45, ha='right')
plt.show()
```

```python
pairplot_cols = ['Age', 'Total Revenue', 'Churn Score', 'CLTV', 'Churn Label']

g= sns.pairplot(
    df[pairplot_cols].dropna(),
    hue='Churn Label',
)
g.fig.suptitle('Pairwise Feature Relationships by Churn Status', y=1.02, fontsize=14, fontweight='bold')
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(11,9))

sns.boxenplot(
    data=df,
    x='Satisfaction Score',
    y='Total Revenue',
    hue='Churn Label'
)

plt.title('Distribution of Total Revenue by satisfaction score & churn')
plt.xlabel('satisfaction score (1 to 5)')
plt.ylabel('total revenue ($)')
plt.legend(title='Churned')
plt.grid(axis='y')

```

```python
import seaborn as sns
import matplotlib.pyplot as plt

g = sns.FacetGrid(
    df,
    row='Senior Citizen',
    col='Gender',
    hue='Churn Label',
    height=3.5,
    aspect=1.3,
    margin_titles=True
)

g.map(sns.scatterplot, 'Age', 'Total Revenue', alpha=0.6)

g.add_legend(title='Churned')
g.fig.subplots_adjust(top=0.9)
g.fig.suptitle('Age vs. Total Revenue Faceted by Demographics')

plt.show()
```

Parameterized SQL queries
Basically we designed parameteres in python bcz sometime the parameter of any result keep changing
suppose that for today the manager demanded i need the customers details who's revenue is minimum than 5000, so you start writing the query and give him the list,     But a day after he wants the customer details with minimum 1000 , so instead of writing the wholw query again , you can simply change the number from 5000  to 1000 outside the main query.

```python
min_revenue = 5000

query = text("""
SELECT
    Customer_ID,
    total_revenue
    FROM telco_clean
    WHERE total_revenue > :min_revenue
    ORDER BY total_revenue DESC
    """)

df = pd.read_sql(
    query,
    engine,
    params={"min_revenue": min_revenue}
)
df.head(10)
```

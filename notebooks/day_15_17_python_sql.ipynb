# Task 3 - Python + SQL Integration

## Objective
Integrate python with MySQL using SQLAlchemy and Pymysql,
execute SQL queries froM Jupyter notebook, and load results
into pandas dataframes.

```python
import sys
sys.path.insert(0, "..")

import pandas as pd
from sqlalchemy import create_engine
from db_utils import get_engine

print("Pandas:", pd.__version__)
print("SQLAlchemy imported successfully")
print("db_utils imported successfully")
```

```python
engine = get_engine()

print("SQLAlchemy engine created successfully!")
```

```python
with engine.connect() as connection:
    print("MySQL database connected successfully!")
```

```python
query = """
SELECT
    Customer_ID,
    total_revenue
FROM telco_clean
ORDER BY total_revenue DESC
LIMIT 10;
"""
```

```python
from sqlalchemy import text

with engine.connect() as connection:
    result = connection.execute(text(query))

    for row in result:
        print(row)
```

```python
df = pd.read_sql(query, engine)
```

```python
df
```

```python
print("Rows:", len(df))
print("Columns:", df.columns.tolist())
```

```python
min_revenue = 5000
```

```python
query + """
SELECT
Customer_ID,
total_revenue
FROM telco_clean
WHERE total_revenue > :min_revenue
ORDER BY total_revenue DESC;
"""
```

```python
df = pd.read_sql(
    query,
    engine,
    params={"min_revenue": min_revenue}
)
df.head(10)
```
